```python

!pip install -q gradio python-docx

import gradio as gr
from docx import Document

def generate_text(name, topic):
    text = f"""
ЗВІТ

Студент: {name}
Тема роботи: {topic}

Робота виконана та здана.
"""
    return text

def generate_doc(name, topic):
    doc = Document()

    text = generate_text(name, topic)

    doc.add_heading("Звіт", 0)
    doc.add_paragraph(text)

    file_path = "result.docx"
    doc.save(file_path)

    return file_path

with gr.Blocks() as demo:
    gr.Markdown("#Генератор документів за шаблоном")

    name = gr.Textbox(label="Ім'я студента")
    topic = gr.Textbox(label="Тема роботи")

    output = gr.File(label="Скачати документ")

    btn = gr.Button("Згенерувати")

    btn.click(
        fn=generate_doc,
        inputs=[name, topic],
        outputs=output
    )

demo.launch(share=True)
```

```python
import re
import requests

url = "https://raw.githubusercontent.com/brown-uk/corpus/master/data/atom/f/f01.xml"
response = requests.get(url)
text = response.text

clean_text = re.sub(r'<[^>]+>', ' ', text)

ukr_words = re.findall(r'\b[А-ЩЬЮЯҐЄІЇа-щьюяґєії\']+\b', clean_text)

all_words = re.findall(r'\b[a-zA-ZА-ЩЬЮЯҐЄІЇа-щьюяґєії\']+\b', clean_text)

numeric_patterns = [
    r'\d+(?:[.,]\d+)?%?',
    r'\b(?=[MDCLXVI])M{0,4}(CM|CD|D?C{0,3})(XC|XL|L?X{0,3})(IX|IV|V?I{0,3})\b'
]
combined_pattern = '|'.join(numeric_patterns)
numeric_values = re.findall(combined_pattern, clean_text)

print(f"--- Статистика по файлу ---")
print(f"Кількість українських слів:    {len(ukr_words)}")
print(f"Загальна кількість слів:       {len(all_words)}")
print(f"Кількість числових значень:    {len(numeric_values)}")
```
